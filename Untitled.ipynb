Using software of your choice, plot the curve where the non-relativistic electron degeneracy pressure \(P_e\) is equal to the pressure of an ideal gas on a log–log scale, with:

- **x-axis:** density (in units of kg m\(^{-3}\))  
- **y-axis:** temperature (in units of K)

Assume:

- mean molecular weight \(\mu = 1\)  
- \(\rho = \mu m_p n_e\)

On the plot, mark and label:

- The position of the Sun’s core: \(\rho_c = 1.5 \times 10^6\) kg m\(^{-3}\), \(T_c = 1.6 \times 10^7\) K  
- A white dwarf: \(\rho = 2 \times 10^9\) kg m\(^{-3}\), central temperature \(8 \times 10^7\) K  

For full credit, upload the file or script used to generate the plot.  


```python
import numpy as np
import matplotlib.pyplot as plt

# Physical constants
kB = 1.380649e-23       # Boltzmann constant, J/K
m_p = 1.6726219e-27     # proton mass, kg
h = 6.62607015e-34      # Planck constant, J*s
me = 9.10938356e-31     # electron mass, kg
pi = np.pi

# Mean molecular weight
mu = 1

# Density array (kg/m^3), logarithmic spacing
rho = np.logspace(3, 10, 500)

# Non-relativistic electron degeneracy pressure formula
# P_e = (h^2 / 5m_e) * (3/(8pi))^(2/3) * n_e^(5/3)
# n_e = rho / (mu * m_p)
n_e = rho / (mu * m_p)
P_e = (h**2 / (5 * me)) * (3 / (8 * pi))**(2/3) * n_e**(5/3)

# Ideal gas pressure: P_ideal = rho * kB * T / (mu * m_p)
# Solve for T where P_ideal = P_e
T = P_e * mu * m_p / (rho * kB)

# Sun's core
rho_sun = 1.5e6   # kg/m^3
T_sun = 1.6e7     # K

# White dwarf
rho_wd = 2e9      # kg/m^3
T_wd = 8e7        # K

# Plot
plt.figure(figsize=(8,6))
plt.loglog(rho, T, label="Electron Degeneracy Pressure = Ideal Gas Pressure", color='blue')

plt.scatter(rho_sun, T_sun, color='orange', s=80, label="Sun's Core")
plt.scatter(rho_wd, T_wd, color='green', s=80, label="White Dwarf")

plt.xlabel(r'Density $\rho$ [kg/m$^3$]')
plt.ylabel(r'Temperature $T$ [K]')
plt.title('Curve: Electron Degeneracy Pressure = Ideal Gas Pressure')
plt.grid(True, which='both', ls='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
